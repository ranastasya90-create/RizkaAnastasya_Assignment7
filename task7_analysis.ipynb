```python
import geopandas as gpd
import pandas as pd
```

1. Load Hasil Task 6

```python
path_input = "../data/input/Analisis_Risiko_Banjir_Terhadap_Tutupan_Lahan_Jaksel.geojson"

gdf = gpd.read_file(path_input)

print("Data berhasil dimuat")
gdf.head()
```

2. Filter Zona Prioritas dengan Skor Tinggi (>= 4)

```python
gdf_prioritas = gdf[gdf['SKOR_TOTAL'] >= 4].copy()

print("Jumlah zona prioritas:", len(gdf_prioritas))
gdf_prioritas.head()
```

3. Hitung Luas dan Ranking

```python
# Ubah ke CRS meter untuk hitung luas
gdf_prioritas = gdf_prioritas.to_crs(epsg=3857)

# Hitung luas
gdf_prioritas['LUAS_M2'] = gdf_prioritas.geometry.area

# Ranking berdasarkan skor & luas
gdf_prioritas = gdf_prioritas.sort_values(
    ['SKOR_TOTAL', 'LUAS_M2'],
    ascending=[False, False]
)

gdf_prioritas.head()
```

4. Ambil Top 5 Zona Mitigasi

```python
zona_top5 = gdf_prioritas.head(5)

zona_top5
```

```python
# Kembalikan ke WGS84
zona_top5 = zona_top5.to_crs(epsg=4326)

zona_top5.head()
```

5. Export ke folder Output

```python
path_output = "../data/output/Zona_Top5_Prioritas.geojson"

zona_top5.to_file(path_output, driver="GeoJSON")

print("File berhasil disimpan di:", path_output)
```
